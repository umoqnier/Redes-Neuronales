## Objetivo del laboratorio

- Practicar lo que vean en clase :)
    - Profundizar en uso de herramientas para desarrollo de soluciones utilizando redes neuronales
- Programar elementos fundamentales de las redes neuronales como los que se listan:
    - Neuronas artificiales
    - Redes profundas
    - Estrategias de optimización 
        - Basadas en gradiente por descenso
        - Basadas en retro-propagación 
- Aplicar los conocimientos vistos para solucionar un problema complejo con base en datos no estructurados
